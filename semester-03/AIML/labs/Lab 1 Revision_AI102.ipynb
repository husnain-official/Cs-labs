```python
def add_all(*numbers):
    return sum(numbers)
print(add_all(1, 2, 3, 4))
```

```python
def show_info(**info):
    for key, value in info.items():
        print(key, value)
show_info(name="Ali", age=21)
```

```python
def factorial(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result
print(factorial(5))
```

```python
def count_vowels(text):
    count = 0
    for ch in text.lower():
        if ch in "aeiou":
            count += 1
    return count
print(count_vowels("Programming"))
```

```python
def reverse_text(text):
    return text[::-1]
print(reverse_text("Python"))
```

```python
def average(numbers):
    return sum(numbers) / len(numbers)
print(average([10, 20, 30]))
```

```python
def find_max(numbers):
    return max(numbers)
print(find_max([5, 9, 2, 7]))
```

## 6. Object-Oriented Programming

```python
class Student:
    pass
s1 = Student()
print(s1)
```

```python
class Student:
    def __init__(self, name):
        self.name = name
s1 = Student("Ali")
print(s1.name)
```

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
s1 = Student("Ali", 21)
print(s1.name, s1.age)
```

```python
class Student:
    def __init__(self, name):
        self.name = name
    def greet(self):
        print("Hello", self.name)
s1 = Student("Sara")
s1.greet()
```

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width
    def area(self):
        return self.length * self.width
r = Rectangle(5, 4)
print(r.area())
```

```python
class BankAccount:
    def __init__(self, balance):
        self.balance = balance
    def deposit(self, amount):
        self.balance += amount
account = BankAccount(1000)
account.deposit(500)
print(account.balance)
```

```python
class Person:
    species = "Human"
    def __init__(self, name):
        self.name = name
p = Person("Ali")
print(p.name)
print(p.species)
```

```python
class Counter:
    count = 0
    def __init__(self):
        Counter.count += 1
c1 = Counter()
c2 = Counter()
print(Counter.count)
```

```python
class Animal:
    def speak(self):
        print("Animal sound")
class Dog(Animal):
    def speak(self):
        print("Bark")
d = Dog()
d.speak()
```

```python
class Cat(Animal):
    def speak(self):
        print("Meow")
animals = [Dog(), Cat()]
for animal in animals:
    animal.speak()
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.__salary = salary
    def get_salary(self):
        return self.__salary
e = Employee("Ali", 50000)
print(e.get_salary())
```

```python
class Shape:
    def area(self):
        raise NotImplementedError
class Square(Shape):
    def __init__(self, side):
        self.side = side
    def area(self):
        return self.side ** 2
print(Square(5).area())
```

## 11. Integrated Practice

```python
students = [{"name": "Ali", "marks": 85}, {"name": "Sara", "marks": 92}, {"name": "Hamza", "marks": 67}]
for s in students:
    print(s["name"], s["marks"])
```

```python
for s in students:
    if s["marks"] >= 80:
        print(s["name"], "Excellent")
```

```python
def grade(mark):
    if mark >= 80:
        return "A"
    elif mark >= 70:
        return "B"
    elif mark >= 60:
        return "C"
    return "F"
print([grade(s["marks"]) for s in students])
```

```python
import pandas as pd
df = pd.DataFrame(students)
print(df)
```

```python
df["Grade"] = df["marks"].apply(grade)
print(df)
```

```python
print(df[df["marks"] >= 70])
```

```python
numbers = [12, 5, 8, 20, 3]
result = {"max": max(numbers), "min": min(numbers), "average": sum(numbers)/len(numbers)}
print(result)
```

```python
def word_count(text):
    words = text.split()
    return len(words)
print(word_count("Python is easy to learn"))
```

```python
def unique_words(text):
    return set(text.lower().split())
print(unique_words("Python python data data science"))
```

```python
class Course:
    def __init__(self, name, students):
        self.name = name
        self.students = students
    def count(self):
        return len(self.students)
course = Course("Python", ["Ali", "Sara", "Hamza"])
print(course.name, course.count())
```

```python
scores = np.array([70, 85, 90, 60, 95])
print("Average:", scores.mean())
```
