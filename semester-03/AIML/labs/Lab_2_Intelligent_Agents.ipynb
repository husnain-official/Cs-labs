## 1. What Is an Intelligent Agent?

An **intelligent agent** is a system that:

- **Perceives** its environment through sensors.
- **Processes** the percepts using an agent program.
- **Acts** on the environment through actuators.

### Basic Agent Structure

```text
Environment
    ↓
  Sensors
    ↓
  Percepts
    ↓
Agent Program
    ↓
  Actions
    ↓
  Actuators
    ↓
Environment
```

The job of AI is to design an **agent program** that implements the agent function:

> **Agent = Architecture + Program**

- **Architecture:** The physical or computational platform on which the agent runs.
- **Program:** The logic that maps percepts to actions.

### Example

For a robot vacuum cleaner:

| Component | Example |
|---|---|
| Environment | Rooms and floor |
| Sensors | Dirt sensor, location sensor |
| Percept | “Dirt detected” |
| Agent program | If dirt is detected, vacuum |
| Actuator | Vacuum motor, wheels |
| Action | Clean or move |

## 2. Agent Programs

An agent program receives the **current percept** from the sensors and returns an action.

There is a difference between an **agent function** and an **agent program**:

- **Agent function:** Maps the complete percept history to an action.
- **Agent program:** Usually receives the current percept and may use stored information, called **internal state**, to make a decision.

# Part A: Simple Reflex Agent

## 3. Simple Reflex Agent

A **simple reflex agent** selects an action using only the **current percept**. It does not remember previous percepts.

It follows **condition-action rules**, also called **if-then rules**.

### General Structure

```text
IF condition is true
THEN perform action
```

### Example Rules for a Room-Cleaning Agent

| Current Percept | Action |
|---|---|
| Dirt detected | Clean |
| No dirt and room is hot | Turn on fan |
| No dirt and room is comfortable | Stay idle |

### Important Point

A simple reflex agent does **not** maintain an internal model of the world. Its decision depends only on what it observes at that moment.

## Example 1: Smart Room Assistant

In this example, we will create a simple reflex agent that decides what to do based on the current room conditions.

The agent receives three percepts:

- `dirt`: Whether dirt is detected.
- `temperature`: The current room temperature.
- `motion`: Whether motion is detected.

The agent will use predefined condition-action rules to select an action.

````python
# Simple Reflex Agent: Smart Room Assistant

def smart_room_agent(dirt, temperature, motion):
    """
    Select an action using only the current percepts.

    Parameters:
        dirt (str): "yes" or "no"
        temperature (float): Room temperature in Celsius
        motion (str): "yes" or "no"

    Returns:
        str: Action selected by the agent
    """

    if dirt.lower() == "yes":
        return "Start cleaning"

    elif temperature > 28:
        return "Turn on the fan"

    elif motion.lower() == "yes":
        return "Turn on the lights"

    else:
        return "Stay idle"


# Collect current percepts
dirt_input = input("Is dirt detected? (yes/no): ")
temperature_input = float(input("Enter room temperature (°C): "))
motion_input = input("Is motion detected? (yes/no): ")

# Agent selects an action
action = smart_room_agent(
    dirt_input,
    temperature_input,
    motion_input
)

print(f"\nSelected Action: {action}")
````

# Part B: Model-Based Reflex Agent

## 4. Model-Based Reflex Agent

A **model-based reflex agent** maintains an **internal state**. This state stores information about the world that may not be directly observable from the current percept.

The agent uses:

1. The current percept.
2. The previous internal state.
3. A model of how the environment changes.

### Why Is Internal State Needed?

Imagine a robot vacuum cleaner that enters a room and cannot currently see whether the room was cleaned earlier.

A simple reflex agent may only react to the current sensor reading.

A model-based agent can remember:

```text
Room A was already cleaned.
Room B has not been cleaned.
```

This stored information helps the agent make a more informed decision.

### General Structure

```text
Current Percept ───────┐
                       ↓
Previous Internal State → Update State → Agent Rules → Action
                       ↑
                Model of the World
```

### Main Difference

| Simple Reflex Agent | Model-Based Reflex Agent |
|---|---|
| Uses current percept only | Uses current percept and internal state |
| No memory of previous situations | Maintains memory/internal state |
| Suitable for fully observable situations | Useful when some information is hidden or unavailable |
| Uses condition-action rules | Uses condition-action rules plus state updates |

## Model-Based Room Monitoring Agent

We will build an agent that monitors a room.

The agent receives:

- `room`: The room currently observed.
- `dirt_detected`: Whether dirt is currently detected.

The agent maintains an internal model containing:

- Which rooms have been cleaned.
- Which rooms are still dirty.
- The last room visited.

### How It Works

1. The agent observes the current room.
2. It updates its internal state.
3. If dirt is detected, it cleans the room.
4. If no dirt is detected, it remembers that the room is clean.
5. It uses its stored state to report the cleaning status of all rooms.

This is model-based because the agent maintains information beyond the current percept.

````python
# Model-Based Reflex Agent: Room Monitoring System

class RoomMonitoringAgent:
    def __init__(self, rooms):
        """
        Initialize the agent's internal state.

        The model stores the cleaning status of each room.
        """
        self.room_status = {}

        for room in rooms:
            self.room_status[room] = "Unknown"

        self.last_room = None

    def update_state(self, room, dirt_detected):
        """
        Update the internal state using the current percept.
        """
        self.last_room = room

        if dirt_detected.lower() == "yes":
            self.room_status[room] = "Dirty"
        else:
            self.room_status[room] = "Clean"

    def choose_action(self, room, dirt_detected):
        """
        Update the model and select an action.
        """
        self.update_state(room, dirt_detected)

        if dirt_detected.lower() == "yes":
            return f"Clean {room}"

        return f"{room} is already clean. No cleaning required."

    def show_internal_state(self):
        """
        Display the information stored in the internal model.
        """
        print("\nInternal State / World Model")
        print("----------------------------")

        for room, status in self.room_status.items():
            print(f"{room}: {status}")

        print(f"Last room visited: {self.last_room}")


# Initialize the agent with a list of rooms
rooms = ["Kitchen", "Bedroom", "Living Room"]
agent = RoomMonitoringAgent(rooms)

# Simulate observations from different rooms
print(agent.choose_action("Kitchen", "yes"))
agent.show_internal_state()

print("\n" + "-" * 40)

print(agent.choose_action("Bedroom", "no"))
agent.show_internal_state()

print("\n" + "-" * 40)

print(agent.choose_action("Living Room", "yes"))
agent.show_internal_state()
````

The dataset contains:

````python
import pandas as pd

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"

columns = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width",
    "species"
]

iris_data = pd.read_csv(url, names=columns)

iris_data.head()
````
